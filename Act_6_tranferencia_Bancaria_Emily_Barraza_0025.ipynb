```python
# =========================================
# SISTEMA DE IA PARA TRANSFERENCIA BANCARIA
# Nivel: Principiante
# =========================================

# Variables iniciales
saldo = 10000
sistema_activo = True
transferencias = 0

print("======================================")
print("   IA DE TRANSFERENCIA BANCARIA")
print("======================================")

# Ciclo principal
while sistema_activo:

    print("\nSaldo disponible:", saldo)
    print("1. Realizar transferencia")
    print("2. Consultar saldo")
    print("3. Ver transferencias realizadas")
    print("4. Ayuda del sistema")
    print("5. Salir")

    opcion = input("Seleccione una opción: ")

    # =====================================
    # OPCIÓN 1: TRANSFERENCIA
    # =====================================
    if opcion == "1":

        cuenta = input("Ingrese la cuenta destino: ")
        monto = float(input("Ingrese el monto a transferir: "))

        # Validaciones
        if monto <= 0:
            print("IA ALERTA: El monto debe ser mayor a 0.")

        elif monto > saldo:
            print("IA ALERTA: Fondos insuficientes.")

        else:

            # Detección de riesgo
            if monto > 5000:

                print("IA DETECCIÓN: Transferencia de alto riesgo.")
                clave = input("Ingrese clave de seguridad: ")

                if clave == "1234":

                    print("IA SEGURIDAD: Clave correcta.")

                    saldo = saldo - monto
                    transferencias = transferencias + 1

                    print("Transferencia realizada con éxito.")
                    print("Cuenta destino:", cuenta)
                    print("Nuevo saldo:", saldo)

                else:
                    print("IA BLOQUEO: Clave incorrecta.")

            else:

                saldo = saldo - monto
                transferencias = transferencias + 1

                print("Transferencia realizada con éxito.")
                print("Cuenta destino:", cuenta)
                print("Nuevo saldo:", saldo)

    # =====================================
    # OPCIÓN 2: CONSULTAR SALDO
    # =====================================
    elif opcion == "2":

        print("\n===== CONSULTA DE SALDO =====")
        print("Saldo actual:", saldo)

    # =====================================
    # OPCIÓN 3: VER TRANSFERENCIAS
    # =====================================
    elif opcion == "3":

        print("\n===== REPORTE DE TRANSFERENCIAS =====")
        print("Transferencias realizadas:", transferencias)

    # =====================================
    # OPCIÓN 4: AYUDA
    # =====================================
    elif opcion == "4":

        print("\n===== AYUDA DEL SISTEMA =====")
        print("1 = Realizar transferencia")
        print("2 = Consultar saldo")
        print("3 = Ver transferencias")
        print("4 = Ayuda")
        print("5 = Salir")

    # =====================================
    # OPCIÓN 5: SALIR
    # =====================================
    elif opcion == "5":

        sistema_activo = False

        print("\n================================")
        print("  REPORTE FINAL DEL SISTEMA")
        print("================================")
        print("Transferencias exitosas:", transferencias)
        print("Saldo final:", saldo)
        print("Cerrando sistema...")

    # =====================================
    # OPCIÓN INVÁLIDA
    # =====================================
    else:

        print("IA ERROR: Opción inválida.")

print("\nPrograma finalizado.")
```
